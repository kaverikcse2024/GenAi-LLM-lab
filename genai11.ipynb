```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from diffusers import StableDiffusionPipeline
from gtts import gTTS
import torch


# Topic
topic = "The benefits of renewable energy"


# =========================================================
# 1. TEXT GENERATION
# =========================================================

print("Generating text...")

text_model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(
    text_model_name
)

text_model = AutoModelForSeq2SeqLM.from_pretrained(
    text_model_name
)

text_prompt = f"Write a short, engaging paragraph about: {topic}"

inputs = tokenizer(
    text_prompt,
    return_tensors="pt"
)

outputs = text_model.generate(
    **inputs,
    max_new_tokens=80
)

generated_text = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("\nGenerated Text:")
print(generated_text)


# =========================================================
# 2. IMAGE GENERATION
# =========================================================

if not torch.cuda.is_available():
    raise RuntimeError(
        "GPU is not available. Go to Runtime > Change runtime type > T4 GPU."
    )

print("\nGenerating image...")

image_prompt = f"An illustration representing {topic}, digital art"

sd_pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)

sd_pipe = sd_pipe.to("cuda")

image = sd_pipe(
    image_prompt,
    num_inference_steps=25
).images[0]

image.save("content_image.png")

print("Image saved as content_image.png")


# =========================================================
# 3. AUDIO GENERATION
# =========================================================

print("\nGenerating audio...")

tts = gTTS(
    text=generated_text,
    lang="en"
)

tts.save("content_audio.mp3")

print("Audio saved as content_audio.mp3")


# =========================================================
# 4. FINAL OUTPUT
# =========================================================

print("\nAll content generated successfully!")
print("Text  : generated")
print("Image : content_image.png")
print("Audio : content_audio.mp3")
```
